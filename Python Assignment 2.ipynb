```python
# Creating two sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print("Set 1:", set1)
print("Set 2:", set2)
```

```python
# Existing sets
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

# Compute union and intersection
union_set = set1.union(set2)
intersection_set = set1.intersection(set2)

print("Union of set1 and set2:", union_set)
print("Intersection of set1 and set2:", intersection_set)
```

```python
# Prompt user for input
score = int(input("Enter your score (0-10): "))

# Conditional statements to categorize performance
if score < 0 or score > 10:
    print("Invalid score! Please enter a value between 0 and 10.")
elif score >= 9:
    print("Performance Category: Excellent")
elif score >= 7:
    print("Performance Category: Good")
elif score >= 5:
    print("Performance Category: Average")
elif score >= 3:
    print("Performance Category: Poor")
else:
    print("Performance Category: Very Poor")
```

```python
# Prompt user for input
score = int(input("Enter your score (0-10): "))

# Determine performance category
if score > 10 or score < 0:
    print("Invalid score! Please enter a value between 0 and 10.")
elif score > 7:
    print("Performance Category: Above Average")
elif score >= 4 and score <= 7:
    print("Performance Category: Average")
else:# score less than 4
    print("Performance Category: Below Average")

```
